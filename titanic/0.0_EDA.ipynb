# Modelo

```python
#DataEncoding facilitado para evitar lentidão de UDF e Pandas.
df_train['Sex_binario'] = df_train['Sex'].map({'female': 1, 'male': 0})

df_test['Sex_binario'] = df_test['Sex'].map({'female': 1, 'male': 0})

#Missing values em todas as variáveis facilitado para teste rápido do modelo simplificado.
df_train = (df_train
           .fillna(-999))

df_test = (df_test
           .fillna(-999))
```

```python
from sklearn.ensemble import RandomForestClassifier 
```

```python
model_classifier = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=0) 

cast_model_vars = ['Sex_binario', 'Age']
```

```python
X = df_train[cast_model_vars]
y = df_train['Survived']
```

```python
model_classifier.fit(X, y)
```

```python
X_prev = df_test[cast_model_vars]
```

```python
p = model_classifier.predict(X_prev)
p
```
